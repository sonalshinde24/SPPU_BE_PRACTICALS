# Weather Data Analysis using MapReduce

This notebook reads data from weather.txt and performs:
- Year-wise temperature analysis
- Max & Min temperature calculation
- Hottest and Coolest year detection

```python
with open("weather.txt", "r") as file:
    data = file.readlines()

print("Sample Data:")
print(data[:5])  # Show first 5 lines
```

```python
def mapper(data):
    mapped = []
    for line in data:
        year, temp = line.strip().split(",")
        mapped.append((year, int(temp)))
    return mapped
```

```python
def reducer(mapped_data):
    result = {}

    for year, temp in mapped_data:
        if year not in result:
            result[year] = []

        result[year].append(temp)

    final = {}

    for year, temps in result.items():
        final[year] = {
            "max": max(temps),
            "min": min(temps)
        }

    return final
```

```python
mapped = mapper(data)
reduced = reducer(mapped)

print("\nYear-wise Temperature Analysis:")
for year, values in reduced.items():
    print(f"{year} -> Max: {values['max']} Min: {values['min']}")
```

```python
hottest_year = max(reduced, key=lambda x: reduced[x]['max'])
coolest_year = min(reduced, key=lambda x: reduced[x]['min'])

print("\nHottest Year:", hottest_year)
print("Coolest Year:", coolest_year)
```
