# Clonal Selection Algorithm (CSA)

This notebook demonstrates:
- Artificial Immune System concept
- Clonal Selection Algorithm
- Optimization using cloning and mutation

```python
import random
```

```python
# Objective function (maximize f(x) = x^2)
def fitness(x):
    return x * x
```

```python
def initialize_population(size):
    return [random.uniform(-10, 10) for _ in range(size)]
```

```python
def clone(population, clone_factor=3):
    clones = []
    for ind in population:
        clones.extend([ind] * clone_factor)
    return clones
```

```python
def mutate(clones, mutation_rate=0.1):
    mutated = []
    for c in clones:
        if random.random() < mutation_rate:
            c += random.uniform(-1, 1)
        mutated.append(c)
    return mutated
```

```python
def select_best(population, size):
    return sorted(population, key=fitness, reverse=True)[:size]
```

```python
def clonal_selection():
    pop_size = 6
    generations = 10

    population = initialize_population(pop_size)

    for gen in range(generations):
        print(f"\nGeneration {gen+1}")

        # Evaluate & Select best
        population = select_best(population, pop_size)
        print("Best:", population[0], "Fitness:", fitness(population[0]))

        # Clone
        clones = clone(population)

        # Mutate
        mutated = mutate(clones)

        # Select next generation
        population = select_best(mutated, pop_size)

    best_solution = max(population, key=fitness)
    print("\nOptimal Solution:", best_solution)
    print("Max Fitness:", fitness(best_solution))
```

```python
clonal_selection()
```

## Explanation

### 🔹 Clonal Selection Algorithm (CSA)
Inspired by the biological immune system.

### 🔹 Steps:
1. Initialize random population (antibodies)
2. Select best individuals based on fitness
3. Clone the selected individuals
4. Apply mutation to introduce variation
5. Select the best again for next generation

### 🔹 Goal:
Maximize the fitness function (here: f(x) = x²)

### 🔹 Outcome:
Algorithm converges to values with higher magnitude (near ±10)
