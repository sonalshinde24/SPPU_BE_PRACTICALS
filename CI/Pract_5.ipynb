# Genetic Algorithm for Parameter Optimization

This notebook demonstrates:
- Population initialization
- Fitness evaluation
- Selection
- Crossover
- Mutation
- Optimization over generations

```python
import numpy as np
import random
```

```python
def nn_error(params):
    # Simulated error function (objective to minimize)
    return sum([p**2 for p in params]) + random.random()
```

```python
population_size = 6
num_generations = 10
mutation_rate = 0.1

# Initial population (each individual has 3 parameters)
population = [np.random.rand(3) for _ in range(population_size)]

print("Initial Population:")
for ind in population:
    print(ind)
```

```python
for gen in range(num_generations):

    # Evaluate fitness
    fitness = [nn_error(ind) for ind in population]

    # Sort population based on fitness (lower is better)
    sorted_pop = [x for _, x in sorted(zip(fitness, population))]

    # Select best 2 individuals (elitism)
    new_population = sorted_pop[:2]

    # Generate new individuals
    while len(new_population) < population_size:
        p1, p2 = random.sample(sorted_pop[:4], 2)

        # Crossover (average)
        child = (p1 + p2) / 2

        # Mutation
        if random.random() < mutation_rate:
            child += np.random.normal(0, 0.1, size=3)

        new_population.append(child)

    population = new_population

    print(f"Generation {gen+1}, Best Error: {min(fitness)}")
```

```python
best_solution = min(population, key=nn_error)
print("\nOptimized Parameters:", best_solution)
```

## Explanation

### 🔹 Fitness Function
A simple function that simulates neural network error.

### 🔹 Selection
Best individuals are selected based on minimum error.

### 🔹 Crossover
Child is created by averaging two parents.

### 🔹 Mutation
Random noise is added to introduce variation.

### 🔹 Result
The algorithm gradually finds optimized parameters with lower error.
